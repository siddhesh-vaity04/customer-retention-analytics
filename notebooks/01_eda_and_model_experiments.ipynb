# Churn Intelligence - Exploratory Data Analysis

This notebook explores the IBM Telco Customer Churn dataset, performing basic EDA and preparing the data for the machine learning pipelines.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Read data (assuming it has been downloaded to data/raw)
df = pd.read_csv('../data/raw/Telco-Customer-Churn.csv')
df.head()
```

## 1. Data Cleaning and Preprocessing

```python
# Convert TotalCharges to numeric, coerce errors to NaN
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
# Fill NaNs with 0 (since tenure is 0 for these records)
df['TotalCharges'].fillna(0, inplace=True)

# Convert Churn to binary
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

df.info()
```

## 2. Target Distribution

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Churn')
plt.title('Churn Distribution')
plt.show()

print("Churn Rate:", df['Churn'].mean())
```

## 3. Numeric Features Analysis

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.histplot(df['tenure'], ax=axes[0], kde=True)
sns.histplot(df['MonthlyCharges'], ax=axes[1], kde=True)
sns.histplot(df['TotalCharges'], ax=axes[2], kde=True)
plt.tight_layout()
plt.show()
```

## 4. Churn by Contract Type

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Contract', y='Churn')
plt.title('Churn Rate by Contract Type')
plt.show()
```

### Conclusions
- Month-to-month contracts have a significantly higher churn rate.
- Higher monthly charges generally correlate with higher churn.
- New customers (low tenure) are more likely to churn.
